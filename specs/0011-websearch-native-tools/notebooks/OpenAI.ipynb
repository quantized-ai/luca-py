### OpenAI Web Search

### Basic mode

Basic mode doesn't include sources or results.

#### Non Streaming

```python
resp = requests.post(
    "https://api.openai.com/v1/responses",
    headers=openai_headers,
    json={
        "model": "gpt-5.4-nano",
        "store": False,
        "input": "Compare Apple’s and NVIDIA’s stock performance today. State each company’s current price and percentage change, explain one reported reason for each movement, and cite a different source for every factual claim.",
        "tools": [{
            "type": "web_search"
        }]
    }
)
resp
```

```python
resp.json()['output']
```

```python
with open('openai_basic_non_streaming_output_2.json', 'w') as fp:
    json.dump(resp.json(), fp, indent=2)
```

#### Streaming

```python
resp = requests.post(
    "https://api.openai.com/v1/responses",
    headers=openai_headers,
    json={
        "model": "gpt-5.4-nano",
        "store": False,
        "input": "Compare Apple’s and NVIDIA’s stock performance today. State each company’s current price and percentage change, explain one reported reason for each movement, and cite a different source for every factual claim.",
        "tools": [{
            "type": "web_search",
            # "external_web_access": False,
        }],
        "stream": True,
    },
    stream=True,
)
resp
```

```python
lines = list(resp.iter_lines(decode_unicode=True))
```

```python
with open('openai_basic_streaming_lines_2.txt', 'w') as fp:
    for line in lines:
        if not line.strip():
            continue
        if line.startswith("data:"):
            line = json.dumps(preview(json.loads(line[6:]), 50), indent=2) + "\n"
        print(line, file=fp)
```

## Open Page action

These examples explore the web search tool action "open_page".

#### Non streaming

```python
prompt = """Search the web for Apple's latest quarterly earnings release.

Then open the official Apple Investor Relations page containing that earnings release and inspect the page itself before answering.

From that page, tell me:
1. The reported quarterly revenue
2. The reported diluted EPS
3. The publication date

Do not answer from search-result snippets alone. You must open the source page and read it."""
```

```python
resp = requests.post(
    "https://api.openai.com/v1/responses",
    headers=openai_headers,
    json={
        "model": "gpt-5.4-nano",
        "store": False,
        "input": prompt,
        "tools": [{
            "type": "web_search"
        }]
    }
)
resp
```

```python
with open('openai_action_open_page_non_streaming_output.json', 'w') as fp:
    json.dump(preview(resp.json()), fp, indent=2)
```
